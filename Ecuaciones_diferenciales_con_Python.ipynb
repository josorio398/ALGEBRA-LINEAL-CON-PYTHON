## **Definimos variables y funciones**

```python
from sympy import *
init_printing()

x = Symbol("x",real= True)
y = Function("y")
```


## **Ejercicio 1: (Ecuación diferencial de primer orden)** 
Resuelve la siguiente ecuación diferencial.

$$y^{\prime} = y$$



```python
Ecu = Eq(diff(y(x),x),y(x))
Ecu 
```

```python
dsolve(Ecu)
```

## **Ejercicio 2:(Ecuación diferencial de segundo orden)** 
Resuelve la siguiente ecuación diferencial.

$$y^{\prime \prime} + y^{\prime} + y =0$$

```python
Ecu = Eq(diff(y(x),x,2)+diff(y(x),x)+y(x),0)
Ecu
```

```python
dsolve(Ecu)
```

## **Ejercicio 3: (Ecuación diferencial con condición inicial)** 
Resuelve la siguiente ecuación diferencial con su condición inicial:

$$\sin(x)\ dx + y\ dy =0, \quad y(0)=1$$

```python
Ecu =Eq(sin(x)+y(x)*diff(y(x),x),0)
Ecu
```

### **Solución general**

```python
dsolve(Ecu)
```

###  **Solución particular**

```python
dsolve(Ecu,ics={y(0):1})
```

## **Tipos de hint para una ecuacion diferencial ordinaria en SymPy**
- **Variables separable:** 'separable'
- **Homogénea:** '1st_homogeneous_coeff_subs_dep_div_indep',    '1st_homogeneous_coeff_subs_indep_div_dep', '1st_homogeneous_coeff_best'.
- **Exacta:** '1st_exact'
- **Lineal:** '1st_linear'
- **Ecuación de Bernoulli:** 'Bernoulli'
- **Écuación de Ricatti:** '1st_rational_riccati'


## **Ejercicio 4: (Ecuación de variables separables - solución explícita)** 
Resuelve la siguiente ecuación diferencial por variable separable:

$$\sec x \ dy= \left(\sec ^2 x+e^{\sin x}\right) \sqrt{1-y^2} \ dx$$

```python
Ecu =Eq(sec(x)*diff(y(x),x),((sec(x))**2+E**sin(x))*sqrt(1-(y(x))**2))
Ecu
```

```python
dsolve(Ecu,hint='separable')
```

## **Ejercicio 5: (Ecuación de variables separables - solución implícita)** 
Resuelve la siguiente ecuación diferencial por variable separable:

$$y(1+\sin x) \ dy=(y-1)^2 \ dx$$

```python
Ecu =Eq(y(x)*(1+sin(x))*diff(y(x),x),(y(x)-1)**2)
Ecu
```

```python
dsolve(Ecu,hint='separable',simplify=False)
```

## **Ejercicio 6: (Ecuación diferencial homogenea)** 
Resuelve la siguiente ecuación diferencial:

$$ (x+2y)\ dx + (2x+3y)\ dy =0$$

```python
Ecu =Eq((x+2*y(x))+(2*x+3*y(x))*diff(y(x),x),0)
Ecu
```

```python
dsolve(Ecu)
```

```python
dsolve(Ecu,hint='1st_homogeneous_coeff_subs_dep_div_indep',simplify=False)
```
