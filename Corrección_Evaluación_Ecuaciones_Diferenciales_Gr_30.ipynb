```python
from sympy import *
init_printing()

x = Symbol("x", real= True)
y = Function("y",real=True)
```

##   **Punto 1: (20 pts)** *(ecuación de variables separables)*

Resuelve la siguiente ecuación diferencial:

$$(1+x) \ dy -y \ dx =0, \quad y(0)=1$$\
$$(1+x) \ \dfrac{dy}{dx} -y  =0, \quad y(0)=1$$

###  **Planteamiento**


```python
edo =Eq((1+x)*diff(y(x),x)-y(x),0)
edo
```


###  **Solución general**

```python
dsolve(edo,hint='separable')
```

###  **Solución particular**

```python
dsolve(edo,hint='separable',ics={y(0):1})
```

##   **Punto 2: (20 pts)** *(ecuación exacta)*

Resuelve la siguiente ecuación diferencial:

$$(\tan{y}-2) \ dx +\left(x\sec^2 y+\dfrac{1}{y} \right) \ dy = 0 $$\
$$(\tan{y}-2)  +\left(x\sec^2 y+\dfrac{1}{y} \right) \ \dfrac{dy}{dx} = 0 $$

###  **Planteamiento**


```python
edo =Eq((tan(y(x))-2)+(x*(sec(y(x)))**2+1/y(x))*diff(y(x),x),0)
edo
```


###  **Solución general**

```python
dsolve(edo,hint='1st_exact',symplify=False).simplify()
```

##   **Punto 3: (20 pts)** *(ecuación homogenea)*

Resuelve la siguiente ecuación diferencial:
$$\left(x+y e^{\frac{y}{x}}\right) \ d x-x e^{\frac{y}{x}} \ d y=0$$\
$$\left(x+y e^{\frac{y}{x}}\right) -x e^{\frac{y}{x}} \ \dfrac{dy}{dx}=0$$

###  **Planteamiento**


```python
edo =Eq((x+y(x)*E**(y(x)/x))-(x*E**(y(x)/x))*diff(y(x),x),0)
edo
```


###  **Solución general**

```python
dsolve(edo,hint='1st_homogeneous_coeff_subs_dep_div_indep',symplify=False).simplify()
```

##   **Punto 4: (30 pts)** 

**Mezclas homogéneas:** en un tanque que contiene 200 galones de salmuera hay 40 libras de sal disueltas. Al tanque ingresa salmuera a razón de tres galones por minuto, y en cada galón hay una libra de sal disuelta. La mezcla, que se mantiene homogénea revolviéndola a gran velocidad, sale del tanque a razón de 1,5 galones por minuto. ¿Qué cantidad de libras de sal hay en el tanque en el instante $t$ ?



$$\frac{dA}{dt} = c_ev_e-\frac{A}{V_i+(v_e-v_s)t}v_s $$


<center> <img src="https://github.com/josorio398/ECUACIONES-DIFERENCIALES-CON-PYTHON/blob/main/imagenes/salmuera.png?raw=true" alt="imafinal" border="0" width ="300"></center>

#### **Definición de funciones, variables y constantes**

```python
t = symbols("t", real= True)
A = Function("A",real=True)
```

#### **Planteamiento ecuación diferencial**

```python
ce = 1    #lb/min
ve = 3    #gal/min
vs = S("3/2")  #gal/min
Vi = 200  #gal

edo = Eq(diff(A(t),t),ve*ce-vs*(A(t)/(Vi+(ve-vs)*t)).simplify())
edo
```

#### **Solución general**

```python
dsolve(edo,hint="1st_linear")
```

#### **Solución particular**

```python
#condiciones iniciales
t0,A0 = 0,40

sol = dsolve(edo,hint="1st_linear",ics={A(t0):A0})
sol
```
